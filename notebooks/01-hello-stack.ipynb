# Hello, semantic stack

Runs inside the `jupyter` service (`make jupyter-up`). Container hostnames work from here: `fuseki`, `qlever`, `oxigraph`.
Load the sample data first with `make load`.

```python
import os
from SPARQLWrapper import SPARQLWrapper, JSON

fuseki = SPARQLWrapper(os.environ.get("FUSEKI_URL", "http://fuseki:3030/ds/sparql"))
fuseki.setReturnFormat(JSON)
fuseki.setQuery(open("data/queries/01-services-by-category.rq").read())
for row in fuseki.query().convert()["results"]["bindings"]:
    print(f'{row["category"]["value"]:<28} {row["service"]["value"]:<22} {row.get("port", {}).get("value", "")}')
```

```python
# Validate the sample data against its SHACL shapes with pySHACL
from rdflib import Graph
from pyshacl import validate

data = Graph().parse("data/sample/semantic-stack.ttl")
shapes = Graph().parse("data/sample/shapes.ttl")
conforms, _, report = validate(data, shacl_graph=shapes)
print("conforms:", conforms)
print(report)
```

```python
# Olympics demo data in QLever: gold medals per team
qlever = SPARQLWrapper(os.environ.get("QLEVER_URL", "http://qlever:7001"))
qlever.setReturnFormat(JSON)
qlever.setQuery(open("data/queries/10-olympics-medals-by-team.rq").read())
import pandas as pd
rows = qlever.query().convert()["results"]["bindings"]
pd.DataFrame([{k: v["value"] for k, v in r.items()} for r in rows])
```
